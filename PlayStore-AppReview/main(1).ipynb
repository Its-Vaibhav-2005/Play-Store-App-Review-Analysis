## Cleaning

```python
playstore = playstore.drop_duplicates()
userreviews = userreviews.drop_duplicates()
```

## Transformation

```python
playstore['Installs'].value_counts()
```

```python
playstore['Installs'] = playstore['Installs'].str.replace("Free", "0").str.replace("+", "", regex=False).str.replace(",", "").astype("float64")
```

```python
playstore['Installs'].value_counts()
```

```python
playstore['Price'].value_counts()
```

```python
playstore['Price'] = playstore['Price'].str.replace("Everyone","0").str.replace("$", "").astype(np.float64)
```

```python
playstore['Price'] = round(playstore['Price']*94.33, 2)
```

```python
playstore['Price'].value_counts()
```

```python
playstore['Reviews'].value_counts()
```

```python
def reviewCount(review):
  try:
    return np.float64(review.replace(",", ""))
  except:
    return np.float64(review.replace("M", ""))*1000000
playstore['Reviews'] = playstore['Reviews'].apply(reviewCount)
```

```python
playstore['Size'].value_counts()
```

```python
def convertSize(size):
  if 'M' in size:
    return np.float64(size.replace("M", ""))
  elif 'k' in size:
    return np.float64(size.replace('k', ""))/1024
  elif 'G' in size:
    return np.float64(size.replace('G', ""))*1000
  else:
    return np.nan
playstore['Size'] = playstore['Size'].apply(convertSize)
playstore['Size'].fillna(playstore['Size'].median(), inplace=True)
```

```python
playstore['Size']
```

## Handling Missing and Droping Irrelavent

```python
# Type - removing the missing rows
# Content Rating - removing the missing rows
# Current Ver / Android Ver - application success doesn't matter on basis of version of application
playstore.dropna(subset=["Type", "Content Rating"], inplace=True)
playstore.drop(columns=["Current Ver", "Android Ver"], axis=1, inplace=True)
```

```python
playstore['Rating'].fillna(playstore['Rating'].mean(), inplace=True)
```

```python
userreviews.dropna(inplace=True)
```

## Finding Insights, Correlation & Trends

### Finding Correlation

```python
playstore.corr(numeric_only=True)
```

```python
px.imshow(
    playstore.corr(numeric_only=True),
    text_auto=True,
    title="Correlation Matrix"
).show()
```

- Review and Installs have strong positive correlation (~0.63) which means more popular apps have more reviews
- Rating has weak correaltion with Installs and review
- Price has negative correlation with Rating, Installs, Reviews which means user prefers free application more over the paids one.

### Category vs Installs

```python
CategoryInstalls = playstore.groupby("Category")['Installs'].mean().sort_values(ascending=False)

px.bar(
    x = CategoryInstalls.index,
    y = CategoryInstalls.values,
    title="Category vs Installs",
).show()
```

- Communication category has the highest average installs, followed by Social and Video Players.
- Categories such as Medical, Events, and Beauty have very low installs.

### Free vs Paid Apps (Box Plot)

```python
px.box(
    playstore,
    x="Type",
    y="Installs",
    title="Category vs Installs",
).show()
```

- Free apps dominate in terms of installs, with significantly higher median and maximum values.
- Paid apps has lower install counts, showing limited reach.
- This suggests that users strongly prefer free applications.

### Rating Distribution

```python
appDistFig = px.histogram(
    playstore,
    nbins=20,
    x="Rating",
    title="Distribution of App Ratings",

)
appDistFig.update_layout(bargap=0.1)
appDistFig.show()
```

- Most apps are rated between 4.0 and 4.5 and very few with 5.
- Very few apps have ratings below 3.0.
- This indicates that the majority of apps maintain a good user experience.

### Size vs Installs

```python
px.scatter(
    playstore,
    x="Size",
    y="Installs",
    log_y=True,
    title="App Size vs Installs",
    opacity=0.5,
    color="Type"
).show()
```

- A high concentration of apps is observed in the lower size range (0–20 MB).
- This indicates that most apps on the Play Store are relatively small in size.

### Size Vs Reviews

```python
px.scatter(
    playstore,
    x="Size",
    y="Reviews",
    log_y=True,
    title="App Size vs Reviews",
    opacity=0.5,
    color="Type"
).show()
```

- Apps across all size ranges (small to large) achieve both low and high install counts.
- There is no consistent upward or downward trend between size and installs.
- Size is not a deciding factor for popularity

### Reviews vs Installs

```python
px.scatter(
    playstore,
    x="Reviews",
    y="Installs",
    trendline="ols",
    title="Reviews vs Installs",
    color="Type",
).show()
```

```python
px.scatter(
    playstore,
    x="Reviews",
    y="Installs",
    log_x=True,
    log_y=True,
    trendline="ols",
    title="Reviews vs Installs (Log Scale)",
    color="Type",
).show()
```
